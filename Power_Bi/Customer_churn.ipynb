## Data Card

This data card provides a description for each column in the `customer_churn_1M.csv` dataset.

| Column Name                   | Description                                                |
| :---------------------------- | :--------------------------------------------------------- |
| `customer_id`                 | Unique identifier for each customer.                       |
| `signup_date`                 | Date and time when the customer signed up.                 |
| `age`                         | Age of the customer.                                       |
| `gender`                      | Gender of the customer (e.g., 'Female', 'Male').           |
| `annual_income`               | Customer's annual income.                                  |
| `education`                   | Customer's highest education level.                        |
| `marital_status`              | Marital status of the customer.                            |
| `dependents`                  | Number of dependents the customer has.                     |
| `tenure`                      | Number of months the customer has been with the company.   |
| `contract`                    | Type of contract the customer has (e.g., 'one_year', 'two_year'). |
| `payment_method`              | Method of payment used by the customer.                    |
| `paperless_billing`           | Indicates if the customer uses paperless billing (0 = No, 1 = Yes). |
| `senior_citizen`              | Indicates if the customer is a senior citizen (0 = No, 1 = Yes). |
| `monthlycharges`              | Monthly charges for the customer's services.               |
| `totalcharges`                | Total charges incurred by the customer over their tenure.  |
| `num_services`                | Number of services the customer is subscribed to.          |
| `has_phone_service`           | Indicates if the customer has phone service (0 = No, 1 = Yes). |
| `has_internet_service`        | Indicates if the customer has internet service (0 = No, 1 = Yes). |
| `has_online_security`         | Indicates if the customer has online security service (0 = No, 1 = Yes). |
| `has_online_backup`           | Indicates if the customer has online backup service (0 = No, 1 = Yes). |
| `has_device_protection`       | Indicates if the customer has device protection service (0 = No, 1 = Yes). |
| `has_tech_support`            | Indicates if the customer has tech support service (0 = No, 1 = Yes). |
| `has_streaming_tv`            | Indicates if the customer has streaming TV service (0 = No, 1 = Yes). |
| `has_streaming_movies`        | Indicates if the customer has streaming movies service (0 = No, 1 = Yes). |
| `customer_satisfaction`       | Customer satisfaction score (typically 1-10).              |
| `num_complaints`              | Number of complaints made by the customer.                 |
| `num_service_calls`           | Number of service calls made by the customer.              |
| `late_payments`               | Number of late payments made by the customer.              |
| `avg_monthly_gb`              | Average monthly gigabytes used by the customer.            |
| `days_since_last_interaction` | Number of days since the customer's last interaction.      |
| `credit_score`                | Customer's credit score.                                   |
| `churn`                       | Target variable: Indicates if the customer churned (0 = No, 1 = Yes). |

```python
service_cols = ['has_phone_service','has_internet_service','has_online_security','has_online_backup','has_device_protection','has_tech_support','has_streaming_tv','has_streaming_movies']

data['services_sum_check'] = data[service_cols].sum(axis=1)

data[['num_services','services_sum_check']].corr()
```

```python
data.drop(columns=['services_sum_check'], inplace=True)
```

```python
tenure = data['tenure_bucket'] = pd.cut(data['tenure'], bins=[0,6,12,24,48,72], labels=['0-6mo','6-12mo','1-2yr','2-4yr','4-6yr'])
tenure
```

```python
data['signup_date'] = pd.to_datetime(data['signup_date'])
```

```python
date=data['signup_year'] = data['signup_date'].dt.year
date
```

```python
income = data['income_bracket'] = pd.cut(data['annual_income'], bins=[0,30000,60000,100000,250000], labels=['Low','Mid','High','Very High'])
income
```

```python
has_complained_flag = data['complaint_flag'] = (data['num_complaints'] > 0).astype(int)
has_complained_flag
```

```python
high_usage_flag = data['high_usage_flag'] = (data['avg_monthly_gb'] > data['avg_monthly_gb'].median()).astype(int)
high_usage_flag
```

```python
from IPython.display import display

churn_rate_new_feature = [
    "tenure_bucket",
    "income_bracket",
    "complaint_flag",
    "contract",
    "high_usage_flag"
]

for col in churn_rate_new_feature:
    print(f"\n📊 Churn Rate by {col}")
    display(data.groupby(col)["churn"].mean().to_frame("Churn Rate"))
```
